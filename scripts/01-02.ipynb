# 1.2. 광전 효과와 빛의 이중성

```python
import numpy as np
from scipy.constants import e

P, r = 100.0, 10.0  # 100W 전구, 10m 거리
I = P / (4 * np.pi * r**2)
a_atom = np.pi * (1e-10) ** 2  # 원자 하나의 단면적
phi = 2.28 * e  # 소듐 일함수

t_wait = phi / (I * a_atom)
print(f"고전 모형의 예측 = {t_wait:.0f} 초")
print(f"실제 실험       < 1e-9 초")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.constants import h

nu = np.linspace(0, 2e15, 1000)
nu0 = 5.51e14
ke = np.where(nu >= nu0, h * (nu - nu0), np.nan)

plt.plot(nu, ke)
plt.title("Maximum kinetic energy of photoelectrons")
plt.xlabel("Frequency (Hz)")
plt.ylabel("Kinetic Energy (J)")
plt.xlim(0, 2e15)
plt.ylim(0, 1e-18)
plt.show()
```
