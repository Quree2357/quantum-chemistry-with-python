# 6.3. 진동과 적외선 분광학

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.constants import u, c


def force_constant(nu, m1, m2):
    mu = m1 * m2 / (m1 + m2) * u
    return mu * (2 * np.pi * c * 100 * nu) ** 2


# (분자, 실측 파수 (cm^-1), 질량1 (u), 질량2 (u), 힘 상수(N/m))
data = [
    ("H2", 4403.56, 1.008, 1.008, 510),
    ("D2", 3116.33, 2.014, 2.014, 527),
    ("HF", 4138.32, 1.008, 18.998, 920),
    ("HCl", 2990.94, 1.008, 35.45, 478),
    ("HBr", 2648.97, 1.008, 79.90, 381),
    ("CO", 2169.81, 12.011, 15.999, 1857),
    ("N2", 2358.57, 14.007, 14.007, 2243),
    ("O2", 1580.19, 15.999, 15.999, 1142),
    ("F2", 916.64, 18.998, 18.998, 454),
    ("Cl2", 559.72, 35.45, 35.45, 319),
]

print(f"{'분자':>2} {'파수':>6} {'mu (u)':>9} {'계산 k':>6} {'실제 k':>6}")
for name, nu, m1, m2, k in data:
    mu = m1 * m2 / (m1 + m2)
    print(f"{name:>3} {nu:8.2f} {mu:8.4f} {force_constant(nu, m1, m2):8.1f} {k:7d}")
```
