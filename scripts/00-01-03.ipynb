# 0.1.3. 행렬과 고유값

```python
import numpy as np

A = np.array([[2.0, 0.0],
              [0.0, 0.5]])
v = np.array([1.0, 1.0])

print("v  =", v)
print("Av =", A @ v)
```

```python
th = np.pi / 4

R = np.array([[np.cos(th), -np.sin(th)],
              [np.sin(th), np.cos(th)]])
v = np.array([1.0, 1.0])

print("Rv =", np.round(R @ v, 6))
print("길이:", np.linalg.norm(v), " -> ", np.linalg.norm(R @ v))
```

```python
A = np.array([[2.0, 1.0],
              [1.0, 2.0]])

v1 = np.array([1.0, 0.0])
v2 = np.array([1.0, 1.0])
v3 = np.array([1.0, -1.0])

for v in [v1, v2, v3]:
    w = A @ v
    print(f"v = {v}  ->  Av = {w}")
```

```python
A = np.array([[2.0, 1.0],
              [1.0, 2.0]])

w, V = np.linalg.eigh(A)

print("고유값:", w)
print("고유벡터(열 방향):\n", np.round(V, 6))
```

```python
A = np.array([[2.0, 1.0],
              [1.0, 2.0]])

w, V = np.linalg.eigh(A)

for i in range(2):
    print(f"Av{i} =", np.round(A@V[:, i], 6))
    print(f"lambda{i} * v{i} =", np.round(w[i] * V[:, i], 6))
```
