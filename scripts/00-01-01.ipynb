```python
import numpy as np

z = 3 + 4j

print("z         =", z)
print("실수부     =", z.real)
print("허수부     =", z.imag)
print("켤레복소수  =", np.conj(z))
print("절댓값     =", abs(z))
```

```python
import matplotlib.pyplot as plt

t = np.linspace(0, 2 * np.pi, 400)

fig, ax = plt.subplots(figsize=(5, 5))
ax.plot(np.exp(1j * t).real, np.exp(1j * t).imag, color="navy", lw=2)
ax.axhline(0, color="gray", lw=0.5)
ax.axvline(0, color="gray", lw=0.5)
ax.set_aspect("equal")
ax.set_xlabel("Re")
ax.set_ylabel("Im")
ax.grid(alpha=0.3)
plt.show()
```

```python
t = np.linspace(0, 2 * np.pi, 400)
th0 = 1.0  # 관찰할 각도 (rad 단위)

fig, (axL, axR) = plt.subplots(
    1, 2, figsize=(10, 3), gridspec_kw={"width_ratios": [1, 2]}
)

# 왼쪽: 복소평면 위의 점과 그 그림자
axL.plot(np.cos(t), np.sin(t), color="navy", lw=1.5)
axL.plot([0, np.cos(th0)], [0, np.sin(th0)], color="crimson", lw=2)
axL.plot(np.cos(th0), np.sin(th0), "o", color="crimson")
axL.plot([np.cos(th0)] * 2, [np.sin(th0), 0], color="gray", ls=":", lw=1.5)
axL.plot(np.cos(th0), 0, "s", color="darkgreen")
axL.set_aspect("equal")
axL.grid(alpha=0.3)
axL.set_xlabel("Re")
axL.set_ylabel("Im")

# 오른쪽: 그림자의 궤적 = 코사인
axR.plot(t, np.cos(t), color="darkgreen", lw=2)
axR.plot(th0, np.cos(th0), "s", color="darkgreen", ms=8)
axR.set_xlabel("theta")
axR.set_ylabel("cos(theta)")
axR.grid(alpha=0.3)
plt.show()
```

```python
x = np.linspace(0, 4 * np.pi, 500)
k = 1.0
A1, phi1 = 1.0, 0.0
A2, phi2 = 1.0, 2 * np.pi / 3

# 방법 1: 실수 코사인끼리 직접 더하기
real_sum = A1 * np.cos(k * x + phi1) + A2 * np.cos(k * x + phi2)

# 방법 2: 복소수로 더한 뒤 마지막에 실수부만 취하기
psi1 = A1 * np.exp(1j * (k * x + phi1))
psi2 = A2 * np.exp(1j * (k * x + phi2))
complex_sum = (psi1 + psi2).real

print("두 방법의 결과가 같은가요?", np.allclose(real_sum, complex_sum))
```

```python
A1, phi1 = 1.0, 0.0
A2, phi2 = 1.0, 2 * np.pi / 3

c = A1 * np.exp(1j * phi1) + A2 * np.exp(1j * phi2)

print(f"합성파 진폭 = {abs(c):.4f}")
print(f"합성파 위상 = {np.angle(c):.4f} rad")
```
