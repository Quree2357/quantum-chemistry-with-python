# 8.1. 수소 원자의 Schrödinger 방정식

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.constants import e, epsilon_0, hbar, m_e

a0 = 5.29177e-11  # 수소 원자의 Bohr 반지름
r = np.linspace(0.02e-10, 8e-10, 500)
V = -(e**2) / (4 * np.pi * epsilon_0 * r)

fig, ax = plt.subplots(figsize=(6, 4))
ax.plot(r * 1e12, V / e, color="black", lw=2)
ax.axhline(0, color="gray", lw=1)
ax.axvline(a0 * 1e12, color="crimson", ls="--", lw=1.5)
ax.text(a0 * 1e12 + 10, -3, "a0 = 52.9 pm", color="crimson", fontsize=11)
ax.set_xlim(0, 500)
ax.set_ylim(-40, 2)
ax.set_xlabel("r (pm)")
ax.set_ylabel("V (eV)")
ax.grid(alpha=0.3)
plt.show()

print(f"r = a0 에서의 퍼텐셜 에너지 = {-(e**2) / (4 * np.pi * epsilon_0 * a0)/e:.4f} eV")
```
