# 7.1. 고리 위 입자

```python
import numpy as np
import matplotlib.pyplot as plt

phi = np.linspace(0, 2 * np.pi, 500)

fig, axes = plt.subplots(1, 3, figsize=(12, 4))

for ax, m in zip(axes, [0, 1, 3]):
    f = 1 / np.sqrt(2 * np.pi) * np.exp(1j * m * phi)

    ax.plot(np.rad2deg(phi), f.real, color="crimson", lw=2, label="Re")
    ax.plot(np.rad2deg(phi), f.imag, color="steelblue", lw=2, label="Im")
    ax.plot(
        np.rad2deg(phi),
        f.real**2 + f.imag**2,
        color="black",
        lw=2,
        ls="--",
        label="|psi|^2",
    )
    ax.axhline(0, color="black", lw=1)
    ax.set_xticks([0, 90, 180, 270, 360])
    ax.set_xlabel("phi (deg)")
    ax.set_ylim(-0.6, 0.6)
    ax.set_title(f"m = {m}")
    ax.grid(alpha=0.3)

axes[0].legend(fontsize=9)
plt.show()
```

```python
from scipy.constants import hbar, m_e, e, h, c

r_benzene = 1.4e-10  # 벤젠의 탄소-탄소 결합 길이: 140 pm
I_benzene = m_e * r_benzene**2
E1 = hbar**2 / (2 * I_benzene) / e
E2 = E1 * 4

print(f"벤젠 분자의 전자가 들뜨기 위해 필요한 에너지: {E2 - E1:.4f} eV")
print(f"해당하는 빛의 파장: {h * c / (E2 - E1) / e * 1e9:.1f} nm")
print(f"실제 벤젠 분자가 흡수하는 빛의 파장: 255 nm")
```
