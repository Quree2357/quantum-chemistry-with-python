# 0.1.4. 대각화와 에르미트 행렬

```python
import numpy as np

A = np.array([[2.0, 1.0],
              [1.0, 2.0]])

w, V = np.linalg.eigh(A)

print("v1 . v1 =", round(np.dot(V[:, 0], V[:, 0]), 10))
print("v1 . v2 =", round(np.dot(V[:, 0], V[:, 1]), 10))
print("\nV^T V =\n", np.round(V.T @ V, 10))
```

```python
D = V.T @ A @ V
print("V^T A V = \n", np.round(D, 10))
```

```python
H = np.array([[2.0, 1.0 - 1.0j], 
              [1.0 + 1.0j, 3.0]])

print("H =\n", H)
print("H와 H의 켤레 전치는 동일한가요?", np.allclose(H, H.conj().T))
```

```python
w2, V2 = np.linalg.eigh(H)

print("고유값:", w2)
print("v1 . v2 =", np.round(np.vdot(V2[:, 0], V2[:, 1]), 10))
```
