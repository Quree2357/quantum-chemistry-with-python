# 9.1. 헬륨 원자

```python
from scipy.constants import physical_constants

Eh = physical_constants["Hartree energy in eV"][0]

Z = 2
E = 2 * (-(Z**2) / 2)

print(f"전자 간의 상호작용을 무시: {E:.2f} hartree = {E * Eh:.2f} eV")
print(f"실험값: {-79.00 / Eh:.2f} hartree = -79.00 eV")
print(f"오차: {abs(E * Eh + 79):.2f} eV ({abs(E * Eh + 79) / 79 * 100:.1f} %)")
```
