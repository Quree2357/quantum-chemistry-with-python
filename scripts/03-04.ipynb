# 3.4. 파동함수가 갖춰야 할 조건

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2, 100)

plt.plot(x, np.exp(5 * x), color="red", lw=2)
plt.xticks([0, 0.5, 1, 1.5, 2])
plt.xlabel("x")
plt.ylabel("u(x)")
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(9, 4))

for ax, (m, ok) in zip(axes, [(2, True), (2.5, False)]):
    phi = np.linspace(0, 2 * np.pi, 100)
    R = 1 + 0.25 * np.cos(m * phi)

    ax.plot(R * np.cos(phi), R * np.sin(phi), color="blue" if ok else "red", lw=2)
    ax.plot(np.cos(phi), np.sin(phi), color="lightgray", ls="--", lw=1)
    ax.plot([1.25], [0], "o", color="black", ms=7)
    ax.plot(
        [1 + 0.25 * np.cos(m * 2 * np.pi)], [0], "o", color="black", ms=7, mfc="white"
    )
    ax.set_aspect("equal")
    ax.axis("off")
    ax.set_xlim(-1.5, 1.5)
    ax.set_ylim(-1.5, 1.5)
    ax.set_title(f"m = {m}", fontsize=12)

plt.tight_layout()
plt.show()
```

```python
L = 1.0
x = np.linspace(0, L, 1000)

f = x * (L - x)
S = np.trapezoid(f**2, x)
A = 1 / np.sqrt(S)

print(f"|f|²의 적분값  = {S:.8f}")
print(f"정규화 상수    = {A:.6f}")
print(f"|Af|²의 적분값 = {np.trapezoid((A * f) ** 2, x):.8f}")
```
