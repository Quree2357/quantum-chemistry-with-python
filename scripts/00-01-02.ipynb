# 0.1.2. 벡터, 내적과 정규직교기저

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("a . b =", np.dot(a, b))
print("직접 계산:", a[0] * b[0] + a[1] * b[1] + a[2] * b[2])
```

```python
print("|a|          =", np.linalg.norm(a))
print("sqrt(a . a)  =", np.sqrt(np.dot(a, a)))
```

```python
theta = np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b))
print(f"a와 b 사이의 각도 = {np.degrees(np.arccos(theta)):.2f}도")
```

```python
v = np.array([1 + 2j, 3 - 1j])

print("켤레 없이 계산 :", np.sum(v * v))
```

```python
print("켤레 씌워서    :", np.sum(np.conj(v) * v))
print("np.vdot(v, v)  :", np.vdot(v, v))
print("|v|^2          :", np.linalg.norm(v) ** 2)
```

```python
th = np.pi / 6  # 30도

e1 = np.array([np.cos(th), np.sin(th)])
e2 = np.array([-np.sin(th), np.cos(th)])

print("e1 . e1 =", round(np.dot(e1, e1), 12))
print("e2 . e2 =", round(np.dot(e2, e2), 12))
print("e1 . e2 =", round(np.dot(e1, e2), 12))
```

```python
v = np.array([3.0, 4.0])

c1 = np.dot(e1, v)
c2 = np.dot(e2, v)
print(f"c1 = {c1:.4f}")
print(f"c2 = {c2:.4f}")

# 계수로 다시 조립해서 원래 벡터가 나오는지 확인
v_rec = c1 * e1 + c2 * e2
print("복원된 v =", np.round(v_rec, 10))
print("원래 v와 같은가요?", np.allclose(v, v_rec))
```

```python
print(f"c1^2 + c2^2 = {c1**2 + c2**2:.6f}")
print(f"|v|^2       = {np.dot(v, v):.6f}")
```
