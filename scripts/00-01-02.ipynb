# 0.1.2. 벡터, 내적과 정규직교기저

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
u = np.array([1, 2, 3])
v = np.array([4, 5, 6])

print("u . v =", np.dot(u, v))
print("직접 계산:", u[0] * v[0] + u[1] * v[1] + u[2] * v[2])
```

```python
print("|u|          =", np.linalg.norm(u))
print("sqrt(u . u)  =", np.sqrt(np.dot(u, u)))
```

```python
cos_theta = np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))
print(f"u와 v 사이의 각도 = {np.degrees(np.arccos(cos_theta)):.2f}도")
```

```python
v = np.array([1 + 2j, 3 - 1j])

print("켤레 없이 계산 :", np.sum(v * v))
```

```python
print("켤레 씌워서    :", np.sum(np.conj(v) * v))
print("np.vdot(v, v)  :", np.vdot(v, v))
print("|v|^2          :", np.linalg.norm(v) ** 2)
```

```python
th = np.pi / 6  # 30도

e1 = np.array([np.cos(th), np.sin(th)])
e2 = np.array([-np.sin(th), np.cos(th)])

print("e1 . e1 =", round(np.dot(e1, e1), 12))
print("e2 . e2 =", round(np.dot(e2, e2), 12))
print("e1 . e2 =", round(np.dot(e1, e2), 12))
```

```python
v = np.array([3.0, 4.0])

c1 = np.dot(e1, v)
c2 = np.dot(e2, v)
print(f"c1 = {c1:.4f}")
print(f"c2 = {c2:.4f}")

# 계수로 다시 조립해서 원래 벡터가 나오는지 확인
v_rec = c1 * e1 + c2 * e2
print("복원된 v =", np.round(v_rec, 10))
print("원래 v와 같은가요?", np.allclose(v, v_rec))
```

```python
print(f"|c1|^2 + |c2|^2 = {np.abs(c1)**2 + np.abs(c2)**2:.6f}")
print(f"|v|^2           = {np.dot(v, v):.6f}")
```
