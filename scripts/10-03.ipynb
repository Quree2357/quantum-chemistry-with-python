# 10.3. 전자 배치와 주기적 경향성

```python
def multiplicity(unpaired_e):
    S = (1 / 2) * unpaired_e
    return int(2 * S + 1)


cases = [("He  1s²", 0), ("Li  [He]2s¹", 1), ("C   [He]2s²2p²", 2), ("N   [He]2s²2p³", 3), ("O   [He]2s²2p⁴", 2)]

names = {1: "singlet", 2: "doublet", 3: "triplet", 4: "quartet"}

for label, n in cases:
    m = multiplicity(n)
    print(f"{label:16s} 홀전자 {n}개 -> S = {0.5*n:.1f}, 스핀 다중도 {m} ({names[m]})")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 1차 이온화 에너지, 출처: NIST Atomic Spectra Database
IE = [
    13.598, 24.587, 5.392, 9.323, 8.298, 11.260, 14.534, 13.618, 17.423,
    21.565, 5.139, 7.646, 5.986, 8.152, 10.487, 10.360, 12.968, 15.760,
    4.341, 6.113, 6.561, 6.828, 6.746, 6.767, 7.434, 7.902, 7.881,
    7.640, 7.726, 9.394, 5.999, 7.900, 9.789, 9.752, 11.814, 14.000,
]

sym = ("H He Li Be B C N O F Ne Na Mg Al Si P S Cl Ar K Ca Sc Ti V Cr Mn Fe Co Ni Cu Zn Ga Ge As Se Br Kr").split()
Z = np.arange(1, 37)

fig, ax = plt.subplots(figsize=(11, 5))
ax.plot(Z, IE, "o-", color="steelblue", lw=2, ms=5)

for z, ie, s in zip(Z, IE, sym):
    if s in ["He", "Ne", "Ar", "Kr"]:
        ax.annotate(s, (z, ie), xytext=(0, 7), textcoords="offset points", ha="center", fontsize=10, color="crimson")
    if s in ["Li", "Na", "K"]:
        ax.annotate(s, (z, ie), xytext=(0, -14), textcoords="offset points", ha="center", fontsize=10, color="seagreen")

ax.set_xlabel("Atomic number Z")
ax.set_ylabel("1st Ionization energy (eV)")
ax.grid(alpha=0.3)
ax.set_xlim(0, 37)
ax.set_ylim(0, 27)
plt.show()
```
