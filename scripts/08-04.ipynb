# 8.4. 스핀과 미세구조

```python
from scipy.constants import physical_constants

alpha = physical_constants["fine-structure constant"][0]
for n in [2, 3]:
    dE = 13.6 * alpha**2 / n**3 / 2
    print(f"n = {n}, l = 1 ({n}p): 약 {dE:.1e} eV")
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(7, 5))

ax.hlines(0, 0.1, 0.9, lw=3, color="navy")
ax.text(0.5, 0.12, "2p  (6-fold)", ha="center", fontsize=10)

ax.hlines(0.35, 1.6, 2.4, lw=3, color="crimson")
ax.text(2.5, 0.35, "j = 3/2   (4-fold)", va="center", fontsize=10)
ax.hlines(-0.7, 1.6, 2.4, lw=3, color="crimson")
ax.text(2.5, -0.7, "j = 1/2   (2-fold)", va="center", fontsize=10)

ax.plot([0.9, 1.6], [0, 0.35], color="gray", ls=":", lw=1.2)
ax.plot([0.9, 1.6], [0, -0.7], color="gray", ls=":", lw=1.2)
ax.set_xlim(0, 3.5)
ax.set_ylim(-1, 1)
ax.set_xticks([])
ax.set_yticks([])
ax.text(0.5, 0.7, "no spin-orbit", ha="center", fontsize=10, color="gray")
ax.text(2.0, 0.7, "with spin-orbit", ha="center", fontsize=10, color="gray")
plt.show()
```
