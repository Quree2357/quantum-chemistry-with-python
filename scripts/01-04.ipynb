# 1.4. 그래서 무엇이 부족했나

```python
from scipy.constants import hbar

L_electron = hbar

m, r, v = 0.05, 0.03, 10
L_top = m * r * v

print(f"n=1 전자의 각운동량 = {L_electron:.3e} J·s")
print(f"고리의 각운동량     = {L_top:.3e} J·s")
print(f"고리를 보어 모형에 넣으면 n = {L_top / hbar:.2e}")
```

```python
E = lambda n: -13.606 / n**2

for n in [1, 2, 10, 50, 100]:
    gap = E(n + 1) - E(n)
    print(f"n={n:4d} -> {n+1:4d}   간격 {gap:.3e} eV   상대 간격 {abs(gap/E(n)):.3e}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(12, 4))

for ax, (lo, hi) in zip(axes, [(1, 6), (10, 15), (50, 55)]):
    ns = np.arange(lo, hi + 1)
    Es = -13.606 / ns**2
    for n, En in zip(ns, Es):
        ax.hlines(En, 0, 1, lw=2, color="navy")
        ax.text(1.04, En, f"{n}", va="center", fontsize=8)
    span = Es.max() - Es.min()
    ax.set_xlim(0, 1.2)
    ax.set_xticks([])
    ax.set_ylim(Es.min() - 0.1 * span, Es.max() + 0.1 * span)
    ax.set_ylabel("energy (eV)")
    ax.set_title(f"n = {lo} ~ {hi}")

plt.tight_layout()
plt.show()
```
