# 0.2.1. 수치미분

```python
import numpy as np

f = np.sin
df = np.cos
x0 = 1.0
exact = df(x0)

print(f"정확한 값         = {exact:.10f}")

for h in [0.1, 0.01, 0.001, 0.0001]:
    d = (f(x0 + h) - f(x0)) / h
    print(f"h ={h:6.0e}: 근사값 = {d:.10f}, 오차 = {abs(d - exact):.3e}")
```

```python
for h in [0.1, 0.01, 0.001, 0.0001]:
    d = (f(x0 + h) - f(x0 - h)) / (2 * h)
    print(f"h ={h:6.0e}: 근사값 = {d:.10f}, 오차 = {abs(d - exact):.3e}")
```

```python
for h in [1e-2, 1e-4, 1e-6, 1e-8, 1e-10, 1e-12]:
    d = (f(x0 + h) - f(x0 - h)) / (2 * h)
    print(f"h ={h:6.0e}: 오차 = {abs(d - exact):.3e}")
```

```python
import matplotlib.pyplot as plt

hs = np.logspace(-15, -1, 400)
err_f = np.abs((f(x0 + hs) - f(x0)) / hs - exact)
err_c = np.abs((f(x0 + hs) - f(x0 - hs)) / (2 * hs) - exact)

fig, ax = plt.subplots(figsize=(8, 5))
ax.loglog(hs, err_f, lw=2, label="forward difference")
ax.loglog(hs, err_c, lw=2, label="central difference")
ax.set_xlabel("h")
ax.set_ylabel("error")
ax.grid(alpha=0.3, which="both")
ax.legend()
plt.show()

print("오차가 가장 작은 h =", hs[np.argmin(err_c)])
```

```python
h = 1e-4
d2 = (f(x0 + h) - 2 * f(x0) + f(x0 - h)) / h**2

print(f"정확한 값  = {-f(x0):.12f}")
print(f"근사값     = {d2:.12f}")
```
